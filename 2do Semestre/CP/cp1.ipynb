```python
class Fecha: 
    def __init__(self, day, month, year):
        self.d: int = day
        self.m: int = month
        self.y: int = year


    def __str__(self):
        return f"{self.d}/{self.m}/{self.y}"
    
hoy = Fecha(3, 11, 2005)
print(hoy)
```

2. Defina una clase `Producto` de una tienda

```python
class Producto:
    def __init__(self, name:str, precio:float, unidad:str):
        self.name:str = name
        self.precio:float = precio
        self.unidad:str = unidad

    def __str__(self):
        return f"el producto {self.name} cuesta {self.precio} por {self.unidad}"
    
producto = Producto("pollo", 300, "lb")
print(producto)
```
