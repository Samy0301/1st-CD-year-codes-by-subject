## 6.1 Resumen

### 6.1.1 Ciclo while
Un ciclo **while** es la forma de ciclo más básica de Python.

`while expresion bool :
  secuencia de instrucciones`

Nos expresa un proceso de repetición que dice que mientras la _expresion bool_ evalue `True` la `secuencia de instrucciones `se ejecuta. Es responsabilidad del programador que la evaluación de la _secuencia de instrucciones_ produzca alguna modificación que haga que en la siguiente repetición la evaluación de la _expresión bool_  de `False`, de lo contrario el ciclo se ejecutaría indefinidamente.

Este tipo de ciclo es suficiente para expresar cualquier proceso de repetición. El lo que sigue vamos a ver otras formas de expresar procesos de repetición que pueden servirnos para facilitar la programación, lograr códigos más legibles o más simples o una ejecución más eficiente, en dependencia del proprósito de la repetición.

### 6.1.2 Ciclo for

El otro tipo de ciclo es el ciclo **for** que sirve para expresar una repetición que pretender "procesar" una "secuencia" de valores.

`for x in secuencia de valores:
  secuencia de instrucciones que suelen hacer algo con x`

Note que ponemos entre comillas los términos _procesar_ y _secuencia_ porque no estamos precisando cómo se produce la secuencia ni qué va a hacer con los datos que produce la secuencia.

### 6.1.3 Ciclo for sobre un rango
Esa secuencia de valores puede expresarse por un **rango** como en el ejemplo siguiente

```python
# from statsmodels.tools.sequences import primes_from_2_to

m = 10
n = 20
for i in range(m, n + 1):
    if i % 3 == 0: print(i)
```

en este caso `range(m, n+1)` nos expresa la secuencia de enteros entre 10 y 20 y el código nos escibe aquellos que son múltiplos de 3.

Cuando queremos que los resultados de procesar la secuencia queden en una lista, el procesamiento de guardarlos en la lista puede expresarse directamente. En el código a continuación la secuencia nos produce enteros que son múltiplos de 3 y los cuadrados de esos valores se guardan directamente en una lista

```python
cuadrados_de_multiplos_de_3 = [i*i for i in range(m, n + 1) if i % 3 == 0]
print(f'cuadrados de los multiplos de 3 entre {m} y {n+1} \n{ cuadrados_de_multiplos_de_3 }')
```

## 6.2 Generadores

¿Cómo construir una función que nos produzca una secuencia de valores para poderlos procesar sin necesidad de guardarlos antes en una lista o en otra estructura contenedora de Python?

En el código a continuación la función `fibonacci_n(n)` se parece a la solución iterativa que estudiamos cuando queríamos calcular el enésimo elemento de la sucesión. Tenemos un ciclo `for _ in range(n)` para devolver el fibonacci calculado y calcular el siguiente. Cuando dentro de este ciclo se ejecuta la instrucción **yield** `a` se interrumpe la ejecución de esta función fibonacci_n y se devuelve el valor (`a `en este ejemplo) al código del `for x in fibonacci_(k)` que procesa este `x `y retorna para seguir ejecutando el código que sigue al **yield** (en este ejemplo para ejecutar `a, b = b, a+b`). Una tal función se denomina en términos de Python un **generador**

Python es quién tras el telón se encarga de sincronizar al _productor_ (el código donde está el **yield**) con el _consumidor_ el código donde se invoca a la función dentro de la cual ocurre el **yield**.

Este patrón **productor-consumidor** intenta independizar la funcionalidad del que _produce_ de la funcionalidad del que _consume_ proucurando con ello más claridad, flexibilidad e idependencia a la hora de hacer modificaciones, es típico de muchas aplicaciones y en particular en la Ciencia de Datos.

### 6.2.1 Generador de generadores
#### 6.2.1.1 Los n primeros de un generador
Un patrón comúnmente utilizado es un generador que produzca un generador **first** a partir de otro generador. Por ejemplo, podemos escribir un generador que a partir de otro generador nos de los n primeros valores producidos por éste.

Nuestro ejemplo de calcular los _n_ primeros de Fibonacci podría escribirse.

```python
def fibonacci_generador():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

def first_n(n, gen):
    """Devuelve los primeros n elementos de gen como generador."""
    i = 0
    for x in gen:
        if i >= n:
            break
        yield x
        i += 1

n = 40
print(f'Los primeros {n} de la sucesion de Fibonacci son')
for x in first_n(n, fibonacci_generador()):
    print(x, end=' ')

lista = [2, 4, 6, 8, 10, 12]
#Una lista se ouede tratar como generador
print()
for x in first_n(4, lista):
    print(x, end=' ')
```

#### 6.2.1.2 Un generador que devuelva los que cumplan una condicion.
Suponemos una condición expresada por una función `condicion` que devuelve un valor bool. La función **where** a continuación va recorriendo los valores que va devolviendo un generador y nos devuelve aquellos que cumplen con la condicion. El código a continuación nos devuelve los elementos de la sucesión de fibonacci que están en un intervalo m,n


```python
def en_intervalo(a, b):
    def check(x):
        return a <= x <= b
    return check

def where(condicion, generador):
    for x in generador:
        if condicion(x): yield x
m=500
n=1000
for k in where(en_intervalo(m,n), fibonacci_generador()):
    print(k, end=' ')
#Estaria infinitamente probando innecesariamente
```

#### 6.2.1.3 Poniendole timeout a un generador
Cuando no tenemos conocimiento de la cantidad de datos que puede producir un generador, y estamos buscando un elemento que cumpla con una condición, puede ser útil poder espefificar alguna forma en que podamos abortar la generación.

El siguiente generador `time_out` pone un límite de tiempo a la generación de un iterador.

```python
import time

def time_out(iterable, max_segundos=60.0):
    start = time.time()
    for x in iterable:
        now = time.time()
        if now - start >= max_segundos:
            return   # termina el generador
        yield x

n = 0.001
for k in time_out(fibonacci_generador(), n):
    pass
print(f'Ultimo fibonacci generado en {n*1000} milisegundos {k}')
```

## 6.3 Iteradores

Intuitivamente podemos decir que en Python un **iterable** es algo que se puede recorrer, y un **iterador** es el objeto instancia de una clase que produce valores. Un iterable se inicia con una operacion `__iter__()` que devuelve un iterador luego va entregando elementos uno a uno con la operación `__next__()` que puede estar dando elementos infinitamente o lanzar la excepcion `StopIteration`.

El protocolo que debe cumplir una clase para que pueda producir objetos que sean "iterables" es que implemente los métodos mágicos siguientes:

```python
class ClaseIterable:
    def __init__(self, datos):
        self.datos = datos

    def __iter__(self):  # ← Este método ES OBLIGATORIO
        return ClaseIterador(self.datos)  # Devuelve un iterador
    # Otros códigos de la clase

class ClaseIterador:
    #...
   def __next__(self, datos):
       # ....devolver algún valor
```

Una misma clase puede iplementar directamente los dos roles si define ambos métodos __init__ y __next__. El código siguiente nos define una clase que vale para generar los números pares menores hasta un valor n

```python
class Pares:
    def __init__(self, maximo):
        self.maximo = maximo

    def __iter__(self):
        self.actual = 0
        return self  # el propio objeto self va a hacer de iterador

    def __next__(self):
        if self.actual > self.maximo:
            raise StopIteration
        par = self.actual
        self.actual += 2
        return par

for k in Pares(10):
    print(k, end=' ')
```

El equivalente tras el telón de la ejeción de un ciclo **for** como el anterior sería en forma más primaria usando **while** es el que se muestra a continuación que enseña todas las "tripas" de la iteración usando` next` y capturando la posible excepcion `StopIteration` para terminar.

```python
# Se convierte internamente en:
iterador = iter(Pares(10))  # Llama a __iter__()
# iterador = Pares(10).__iter__() # Comente el anterior y ejecute este, es lo mismo
while True:
    try:
        k = next(iterador)         # Llama a __next__()
        print(k, end=' ')
    except StopIteration:          # El iterador se agota
        break
```

### 6.3.1 Un iterable para la sucesión de Fibonacci

```python
class FibonacciIterable():
    def __init__(self):
        self.primero = 0
        self.segundo = 1
        self.penultimo = self.primero
        self.ultimo = self.segundo

    def __iter__(self) -> 'FibonacciIterable':
        self.penultimo = self.primero
        self.ultimo = self.segundo
        return self

    def __next__(self) -> int:
        # Simplemente genera Fibonacci infinito
        resultado = self.penultimo
        self.penultimo, self.ultimo = self.ultimo, self.penultimo + self.ultimo
        return resultado  # ¡Nunca para!

class FirstN():
    def __init__(self, it, n): #it debe cumplir el protocolo iterador
        self._it = it
        self._n = n
    def __iter__(self):
        self._faltan = self._n
        return self
    def __next__(self):
        if self._faltan <= 0:
            raise StopIteration #esta excepción detiene un for si esto se está haciendo dentro de un for
        try:
            item = next(self._it)
            self._faltan -= 1
            return item
        except StopIteration:
            raise StopIteration

for x in FirstN(FibonacciIterable(), 10): print(x, end = ' ')
print()
for x in FirstN(FibonacciIterable(), 10): print(x, end = ' ')
print()

#Descomente el código siguiente y pruebe
print()
fib10 = FirstN(FibonacciIterable(), 10)
for x in fib10: print(x, end = ' ')
print()
for x in fib10: print(x, end = ' ')
#¿Por qué los resultados son diferentes?
```

### 6.3.2 ¿Generador o Iterador?
Al usar del **yield** en una función geneneradora se logra una sintaxis más simple que si se usa toda la maquinaria equivalente basada en el protocolo de iteración.  pero con una sintaxis mucho más sencilla. El código a continuación nos muestra nuestro ejemplo de los números pares de las dos formas.

Use **yield** siempre que se exprese su código de modo más sencillo y seguro. Trabaje con el protocolo de iteración cuando quiera definir iteradores más personalizables cuando quiera permitir que el objeto iterador pueda ser modificable entre la aplicación de un **next** y la siguiente

```python
class Pares:
    def __init__(self, maximo): self.maximo = maximo
    def __iter__(self): self.actual = 0; return self
    def __next__(self):
        if self.actual > self.maximo: raise StopIteration
        par = self.actual; self.actual += 2; return par

def pares(maximo):
    actual = 0
    while actual <= maximo:
        yield actual  # ← Pausa aquí, mantiene estado
        actual += 2

for k in Pares(20):
    print(k, end = ' ')
print()

for k in pares(20):
    print(k, end = ' ')
print()
```

## 6.4 Lazy evaluation
La **evaluación perezosa** o **evaluación retardada** (_lazy evaluation_) cuado se recorre una secuencia de valores con un ciclo de la forma

**for** x **in** _Productor_de_Valores_:
  ..._Algún procesamiento sobre_ x ....

Cuando el _productor_de_valores_ es algún tipo de estructura básica de Python como una lista, un diccionario, un tuplo, etc. Los valores ya están guardados en memoria, lo que hace la maquinaria del **for** es irlos tomando directamente de la memoria. Pero, supongamos que los valores son producidos por un procedimiento costoso en tiempo. Cuando se dice evaluación perezosa quiere decir que el ciclo for "manda" a producir un valor solo en el momento en que lo va a usar en cada iteración (no depende de que están guardados en una estructura en memoria) para entonces procesar este valor. De modo que si el ciclo for decide terminar no necesita que se hayan producido todos los valores lo que puede significar una ganancia en tiempo.

En el siguiente ejemplo queremos procesar valores que están en un archivo json el primer código lee completamente el json hacia una lista

```python
import json
import time
import gc

#OPCION LEYENDO A LISTA Y LUEGO PROCESANDO LISTA
ruta_archivo = "imdb_titulos.json"
def buscar_en_lista(ruta_archivo, longitud_minima):
    inicio = time.perf_counter()

    with open(ruta_archivo, 'r', encoding='utf-8') as f:
        datos = [json.loads(linea) for linea in f if linea.strip()]

    encontrado = None
    posicion = -1
    for texto in datos:
        posicion+=1
        if len(texto) > longitud_minima:
            encontrado = texto
            break

    fin = time.perf_counter()
    gc.collect() #Para que no quede nada en memoria de lo cual internamente Python se pudiera aprovechar en la siguiente función
    return encontrado, posicion, fin - inicio


# Ejemplo de uso
print("OPCION 1 Cargando todo a lista ....")
resultado, pos, tiempo = buscar_en_lista(ruta_archivo, 47)
print("Resultado:", resultado)
print(f"{resultado}, longitud {len(resultado)} en la posición {pos} en tiempo: {tiempo:.6f} segundos")

#OPCION LEYENDO DIRECTAMENTE DEL ARCHIVO CON LAZY EVALUACION
class JsonLStringReader:
    def __init__(self, ruta_archivo):
        self.ruta_archivo = ruta_archivo
        self.archivo = None

    def __iter__(self):
        self.archivo = open(self.ruta_archivo, 'r', encoding='utf-8')
        return self

    def __next__(self):
        if self.archivo is None:
            raise ValueError("Hay que iterar desde un for o con un iterador haciendo primero iter().")

        while True:
            linea = self.archivo.readline()
            if not linea:
                self.archivo.close()
                raise StopIteration

            linea = linea.strip()
            if linea:
                return json.loads(linea)

def buscar_lazy(ruta_archivo, longitud_minima):
    inicio = time.perf_counter()

    encontrado = None
    posicion = -1
    for texto in JsonLStringReader(ruta_archivo):
        posicion+=1
        if len(texto) > longitud_minima:
            encontrado = texto
            break

    fin = time.perf_counter()
    return encontrado, posicion, fin - inicio


# Ejemplo de uso
print("\nOPCION 2 Procesando con lazy evaluation ....")
resultado, pos, tiempo = buscar_en_lista(ruta_archivo, 47)
print("Resultado:", resultado)
print(f"{resultado}, longitud {len(resultado)} en la posición {pos} en tiempo: {tiempo:.6f} segundos")
```

Esta medición del tiempo es muy burda, pero es para que tenga una idea. Realmente esto depende del procesador del Python que esté usando, del tamaño del archivo. Para archivos reales de Gigas de tamaño puede ser inviable por falta de memoria cargar todo el archivo a una lista. Piense por ejemplo en una aplicación que va mostrándole sus emails y que les va mostrando los emails de segmentos en segmentos
