## 6.1 Resumen 

### 6.1.1 Ciclo while
Un ciclo **while** es la forma de ciclo más básica de Python.

`while expresion bool :
  secuencia de instrucciones`

Nos expresa un proceso de repetición que dice que mientras la _expresion bool_ evalue `True` la `secuencia de instrucciones `se ejecuta. Es responsabilidad del programador que la evaluación de la _secuencia de instrucciones_ produzca alguna modificación que haga que en la siguiente repetición la evaluación de la _expresión bool_  de `False`, de lo contrario el ciclo se ejecutaría indefinidamente.

Este tipo de ciclo es suficiente para expresar cualquier proceso de repetición. El lo que sigue vamos a ver otras formas de expresar procesos de repetición que pueden servirnos para facilitar la programación, lograr códigos más legibles o más simples o una ejecución más eficiente, en dependencia del proprósito de la repetición.

### 6.1.2 Ciclo for

El otro tipo de ciclo es el ciclo **for** que sirve para expresar una repetición que pretender "procesar" una "secuencia" de valores.

`for x in` _secuencia_ de valores:
  secuencia de instrucciones que suelen _procesar x_

Note que distinguimos los términos _secuencia_ y _procesar_ porque no estamos precisando cómo se produce la secuencia ni qué va a hacer con los datos que produce la secuencia.

### 6.1.3 Ciclo for sobre un rango
Esa secuencia de valores puede expresarse por un **rango** como en el ejemplo siguiente

```python
# from statsmodels.tools.sequences import primes_from_2_to

m = 10
n = 20
for i in range(m, n + 1):
    if i % 3 == 0: print(i)
```

en este caso `range(m, n+1)` nos expresa la secuencia de enteros entre 10 y 20 y el código nos escribe aquellos que son múltiplos de 3.

Cuando queremos que los resultados de procesar la secuencia queden en una lista, el procesamiento de guardarlos en la lista puede expresarse directamente. En el código a continuación la secuencia nos produce enteros que son múltiplos de 3 y los cuadrados de esos valores se guardan directamente en una lista

```python
cuadrados_de_multiplos_de_3 = [i*i for i in range(m, n + 1) if i % 3 == 0]
print(f'cuadrados de los multiplos de 3 entre {m} y {n+1} \n{ cuadrados_de_multiplos_de_3 }')
```

## 6.2 Generadores

¿Cómo construir una función que nos produzca una secuencia de valores para poderlos procesar sin necesidad de guardarlos antes en una lista o en otra estructura contenedora de Python como tuplos o diccionarios?

En el código a continuación la función `fibonacci_n(n)` se parece a la solución iterativa que estudiamos cuando queríamos calcular el enésimo elemento de la sucesión. Tenemos un ciclo `for _ in range(n)` para devolver el fibonacci calculado y calcular el siguiente. Cuando dentro de este ciclo se ejecuta la instrucción **yield** `a` se interrumpe la ejecución de esta función fibonacci_n y se devuelve el valor (`a `en este ejemplo) al código del `for x in fibonacci_(k)` que procesa este `x `y retorna para seguir ejecutando el código que sigue al **yield** (en este ejemplo para ejecutar `a, b = b, a+b`). Una tal función se denomina en términos de Python un **generador**

Python es quién tras el telón se encarga de sincronizar al _productor_ (el código donde está el **yield**) con el _consumidor_ el código donde se invoca a la función dentro de la cual ocurre el **yield**.

Este patrón **productor-consumidor** intenta independizar la funcionalidad del que _produce_ de la funcionalidad del que _consume_ proucurando con ello más claridad, flexibilidad e idependencia a la hora de hacer modificaciones, es típico de muchas aplicaciones y en particular en la Ciencia de Datos.

Si ejecutamos el código de la celda siguiente ejecutará infinitamente. 

Realmente este no ejecutará infinitamente porque dará excepción al calcular un número que soprepase la cantidad de dígitos que la conversión de `int `a `string` permita para hacer en `print(x)`

### 6.2.1 Generador de generadores
#### 6.2.1.1 Los n primeros de un generador
Un patrón comúnmente utilizado es un generador que produzca un generador **first** a partir de otro generador. Por ejemplo, podemos escribir un generador que a partir de otro generador nos de los n primeros valores producidos por éste.

Nuestro ejemplo de calcular los _n_ primeros de Fibonacci podría escribirse.

```python
def fibonacci_generador():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

def first_n(n, gen):
    """Devuelve los primeros n elementos de gen como generador."""
    i = 0
    for x in gen:
        if i >= n:
            break
        yield x
        i += 1

n = 40
print(f'Los primeros {n} de la sucesion de Fibonacci son')
for x in first_n(n, fibonacci_generador()):
    print(x, end=' ')

lista = [2, 4, 6, 8, 10, 12]
#Una lista se puede tratar como generador
print()
for x in first_n(4, lista):
    print(x, end=' ')
```

#### 6.2.1.2 Un generador que devuelva los que cumplan una condicion.
Suponemos una condición expresada por una función `condicion` que devuelve un valor bool. La función **where** a continuación va recorriendo los valores que va devolviendo un generador y nos devuelve aquellos que cumplen con la condicion. El código a continuación nos devuelve los elementos de la sucesión de fibonacci que están en un intervalo m,n

Note cómo hemos construído un generador a partir de otro generador


```python
def en_intervalo(a, b):
    def check(x):
        return a <= x <= b
    return check

def where(condicion, generador):
    for x in generador:
        if condicion(x): yield x
m=500
n=1000
for k in where(en_intervalo(m,n), fibonacci_generador()):
    print(k, end=' ')
#Estaria infinitamente probando innecesariamente, no hay mas numeros de Fibonacci fuera de ese intervalo
#Pero en este ejemplo cómo conocemos el generador es infinito
```

#### 6.2.1.3 Poniendole timeout a un generador
Cuando no tenemos conocimiento de la cantidad de datos que puede producir un generador, y estamos buscando un elemento que cumpla con una condición t no queremo squedarnis cortos con algo como `first_n`, puede ser útil poder especificar alguna forma en que podamos abortar la generación.

El siguiente generador `time_out` pone un límite de tiempo a la generación de un iterador y así evitar caer en un ciclo infinito.

```python
import time

def time_out(iterable, max_segundos=60.0):
    start = time.time()
    for x in iterable:
        now = time.time()
        if now - start >= max_segundos:
            return   # termina el generador
        yield x

n = 0.00005
for k in time_out(fibonacci_generador(), n):
    pass
print(f'Ultimo fibonacci generado en {n*1000} milisegundos {k}')
```

## 6.3 Iteradores

Intuitivamente podemos decir que en Python un **iterable** es algo que se puede recorrer, y un **iterador** es el objeto instancia de una clase que produce valores. Un iterable se inicia con una operacion `__iter__()` que devuelve un iterador luego va entregando elementos uno a uno con la operación `__next__()` que puede estar dando elementos infinitamente o lanzar la excepcion `StopIteration`.

El protocolo que debe cumplir una clase para que pueda producir objetos que sean "iterables" es que implemente los métodos mágicos `__iter__` y `__next__` siguientes:

```python
class ClaseIterable:
    def __init__(self, datos):
        self.datos = datos

    def __iter__(self):  # ← Este método ES OBLIGATORIO
        return ClaseIterador(self.datos)  # Devuelve un iterador
    # Otros códigos de la clase

class ClaseIterador:
    #...
   def __next__(self, datos):
       # ....devolver algún valor
```

Una misma clase puede implementar directamente los dos roles si define ambos métodos __init__ y __next__. El código siguiente nos define una clase que vale para generar los números pares menores hasta un valor n

```python
class Pares:
    def __init__(self, maximo):
        self.maximo = maximo

    def __iter__(self):
        self.actual = 0
        return self  # el propio objeto self va a hacer de iterador

    def __next__(self):
        if self.actual > self.maximo:
            raise StopIteration
        par = self.actual
        self.actual += 2
        return par

for k in Pares(10):
    print(k, end=' ')
```

Note que los números no se han guardado en ninguna lista u otra estructura.

### 6.3.1 Un iterable para la sucesión de Fibonacci

```python
class FibonacciIterable():
    def __init__(self):
        self.primero = 0
        self.segundo = 1
        self.penultimo = self.primero
        self.ultimo = self.segundo

    def __iter__(self) -> 'FibonacciIterable':
        self.penultimo = self.primero
        self.ultimo = self.segundo
        return self

    def __next__(self) -> int:
        # Simplemente genera Fibonacci infinito
        resultado = self.penultimo
        self.penultimo, self.ultimo = self.ultimo, self.penultimo + self.ultimo
        return resultado  # ¡Nunca para!

class FirstN():
    def __init__(self, it, n): #it debe cumplir el protocolo iterador
        self._it = it
        self._n = n
    def __iter__(self):
        self._faltan = self._n
        return self
    def __next__(self):
        if self._faltan <= 0:
            raise StopIteration #esta excepción detiene un for si esto se está haciendo dentro de un for
        try:
            item = next(self._it)
            self._faltan -= 1
            return item
        except StopIteration:
            raise StopIteration

for x in FirstN(FibonacciIterable(), 10): print(x, end = ' ')
print()
for x in FirstN(FibonacciIterable(), 10): print(x, end = ' ')
print()

# #Descomente el código siguiente y pruebe
print()
fib10 = FirstN(FibonacciIterable(), 10)
for x in fib10: print(x, end = ' ')
print()
for x in fib10: print(x, end = ' ')
#¿Por qué los resultados son diferentes?
#El objeto fib10 sigue siendo el mismo de antes se siguen generando otros 10 más a partir de donde se quedó
```

### 6.3.2 ¿Generador o Iterador?
Al usar el **yield** en una función geneneradora se logra una sintaxis más simple que si se usa toda la maquinaria equivalente basada en el protocolo de iteración. El código a continuación nos muestra nuestro ejemplo de los números pares de las dos formas.

Use **yield** siempre que se exprese su código de modo más sencillo y seguro. Trabaje con el protocolo de iteración cuando quiera definir iteradores más personalizables como cuando quiere permitir que el objeto iterador pueda ser modificable entre la aplicación de un **next** y la siguiente

El código a continuación nos muestra *tres* formas de hacer una iteración para procesar los datos producidos por un generador.

La primera es usar el recurso `yield` dentro de una función. La función actúa como un objeto generador. Python ejecuta la maquinaria de iteración por detrás

La segunda es a través de una clase para crear un objeto generador mediante la implementación adecuada de los métodos `__iter__` y `__next__` dentro de la clase. Cuando este generador se usa correctamente en un ciclo `for` Python se encarga de hacer el recorrido por los elementos generados sin que el código tenga que ocuparse explicitamente de usar el `__init__` y el `__next__`. Pero puede alterarse la generación si el código que hace la iteración modifica explícitamente el objeto genrador (vea los comentarios en el código de más abajo)

La tercera opción es la que el mismo código que hace el recorrido y procesamiento de los elementos generados se ocupe explítamente de invocar a los mismos usando un ciclo `while` y capture y trate adecuadamente la excepción que pueda disparar el `__next__` cuando ya no haya más elementos a generar

La primera de las opciones es la más simple, clara y segura porque nos evita cometer errores. La segunda y la tercera nos permiten lograr efectos mas personalizados modificando al objeto generador durante la iteración pero que son mas promensas a que se comentan errores.


```python
import random
class Pares:
    def __init__(self, maximo): self.maximo = maximo
    def __iter__(self): self.actual = 0; return self
    def __next__(self):
        if self.actual > self.maximo: raise StopIteration
        par = self.actual; 
        self.actual += 2; 
        return par

def pares(maximo):
    actual = 0
    while actual <= maximo:
        yield actual  # ← Pausa aquí, mantiene estado para la siguiente iteración
        actual += 2

max = 20
print(f'Iterando un generador yield con for los pares hasta {max}')
for k in pares(max):
    print(k, end = ' ')
print('\n')

pares20 = Pares(20)
print(f'Iterando un objeto iterable con for (iter, next) los pares hasta {max}')
for k in pares20:
    print(k, end = ' ')

    #Descomentar para probar que por el camino puedo alterar la maquinaria
    pares20.maximo=30
print('\n')

pares20.maximo = 7
pares20 = Pares(20)
for k in pares20:
    print(k, end = ' ')
print('\n')

print(f'Iterando explicitamente con while un objeto iterable (iter, next) los pares hasta {max}')
pares20.__iter__() #Preparar a pares20 para que se le aplique __next__()
while True:
  try:
    k=pares20.__next__() #__next__() debe disparar una excepción cuando no se puede seguir aplicando
    print(k, end = ' ') #Procesar el elemento retornado por el __next__()

    #Descomentar para probar que por el camino puedo alterar la maquinaria
    pares20.maximo=10
  except: 
    break #Si no hay más elementos a iterar se aborta el ciclo

  
```

## 6.4 Lazy evaluation

Al recorrer una secuencia de valores con un ciclo de la forma

**for** x **in** _Productor_de_Valores_:
  ..._Algún procesamiento P sobre_ x ....

y el _Productor_de_Valores_ es algún tipo de estructura de datos básica de Python como una lista, un diccionario, un tuplo, etc. Los valores ya están guardados en memoria. Lo que hace Python con la maquinaria del **for** **in** es irlos tomando directamente de la memoria y procesarlos. 

Pero, supongamos que los valores son producidos por un procedimiento costoso en tiempo, y puede que no necesitemos guardarlos en una estructura oara procesarlos después. Cuando se dice **evaluación perezosa** (**lazy evaluation**) quiere decir que el ciclo for "manda" a producir un valor solo en el momento en que ese valor se va a usar en cada iteración (no depende de que están guardados previamente en una estructura en memoria) para entonces procesar __P__ este valor. De modo que si el ciclo **for** decide terminar por alguna razón no necesita que previamente se hayan producido todos los valores. Esto puede significar una ganancia en tiempo.

En el siguiente ejemplo queremos procesar valores que están en un archivo json.

El primer segmento de código lee completamente el json hacia una lista

```python
import json
import time
import gc

a_buscar = "Jungle"
#OPCION 1 LEYENDO A LISTA Y LUEGO PROCESANDO LISTA
ruta_archivo1 = "peliculas.json"
print("Generacion cargando archivo a lista y recorriendo lista haciendo yield...")
inicio = time.perf_counter()
with open(ruta_archivo1, 'r', encoding='utf-8') as f1:
  films = json.load(f1) #Lee toda la lista de una vez
encontro = False
for film in films:
  if a_buscar in film:
    # tiempo = time.perf_counter() - inicio
    encontro = True
    break
tiempo = time.perf_counter() - inicio
if encontro:     
    print(f'{a_buscar} encontrado en film {film} en tiempo {tiempo:.6f} segs')
else: 
   print(f'{a_buscar} NO encontrado  en tiempo {tiempo:.6f} segs')

#OPCION 2 LEYENDO LINEA A LINEA DEL ARCHIVO E IR DEVOLVIENDO CADA LINEA CON yield 
ruta_archivo2 = "peliculas.jsonl"
def generador_leyendo_linea_a_linea(ruta_archivo2):
    inicio = time.perf_counter()
    with open(ruta_archivo2, 'r', encoding='utf-8') as f2:
      for linea in f2:
        linea=linea.strip()
        yield json.loads(linea)

print("\nCargando linea a linea por demanda del archivo jsonl ...")
inicio = time.perf_counter()
for film in generador_leyendo_linea_a_linea(ruta_archivo2):
  if a_buscar in film:
    encontro = True
    break
  else: encontro = False
tiempo = time.perf_counter() - inicio
if encontro:print(f'{a_buscar} encontrado en film {film} en tiempo {tiempo:.6f} segs')
else:
  tiempo = time.perf_counter() - inicio
  print(f'{a_buscar} NO encontrado  en tiempo {tiempo:.6f} segs')
```
