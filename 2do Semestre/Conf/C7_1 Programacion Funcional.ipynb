## 7.1 Qué significa que una función sea un objeto

En muchos lenguajes de programación las funciones ocupan una categoría especial dentro del lenguaje. Pueden invocarse (generalmente mediante el nombre de la función encerrando entre paréntesis los parámetros que se le pasan a la función), pero no pueden manipularse de la misma forma que otros valores.

Python adopta un enfoque diferente. Las funciones en Python pueden considerarse también como objetos. Esto significa las funciones pueden:

•	**Almacenarse (asignarse a) en variables**.

•	**Pasarse como argumento a otra función**.

•	**Almacenarse como valores dentro de estructuras como listas, diccionarios, tuplos, ...**.

•	**Que una función pueda devolver como resultado otra función (función que puede ser creada dentro de su ejecución)**.

•	**"Inspeccionarse" durante la ejecución. Que se puedan consultar propiedades de la función**.

Estas características constituyen uno de los pilares fundamentales de la programación funcional y que puede hacer más simple, expresiva y eficiente la programación en Python.

La siguiente función `kg_a_libras` si se invoca en la forma 

`print(kg_a_libras(70))`

nos devuelve el valor en libras de un peso que se le pasa en kilogramos. Pero su nombre puede usarse como una variable, de modo que `print(type(kg_a_libras))` nos escribiría el tipo de la "variable" `kg_a_libras` y `print(kg_a_libras)` nos escribiría el valor de esa variable que sería en este caso indicar que es una función de nombre `kg_a_libras` más algo así como el lugar en memoria donde está el código de la misma 

```python
def kg_a_libras(peso_en_Kgs):
    return peso_en_Kgs * 2.20462

print(kg_a_libras(70))
print(type(kg_a_libras))
print(kg_a_libras)
```

## 7.2 Asignación a una variable como si fuera un valor

Si una función es un objeto, debería poder almacenarse en una variable exactamente igual que un número, una cadena u otro valor cualquiera.

Si hacemos

`convierte = kg_a_libras`

ahora el valor de la variable `convierte` es la misma función que `kg_a_libras`. Es decir, realmente una referencia al mismo objeto función

print(convierte==kg_a_libras) 

nos debe escribir `True`

Ambas variables *apuntan* al mismo objeto. Compruébelo ejecutando el segmento a continuación

```python
def kg_a_libras(peso_en_Kgs):
    return peso_en_Kgs * 2.20462

convierte = kg_a_libras

print(convierte == kg_a_libras)
kgs = 1.5

#Note que invocados como función dan el mismo valor
print(f'\n{kgs} kgs convertidos a libras con kg_a_libras son {kg_a_libras(kgs)} libras')
print(f'{kgs} kgs convertidos a libras con convierte son   {convierte(kgs)} libras')

#Note que escritos como valores son el mismo valor
print(f'\nkg_a_libras es el objeto {kg_a_libras}')
print(f'covierte es el objeto    {convierte}')
```

## 7.3 Pasar funciones como parámetros a otra función
El siguiente bloque de código define varias funciones de conversión entre unidades de medida

La función `convierte(f, valor)` escribe el resultado de aplicar la función de conversión *`f`* al parámetro *`valor`*` y nos devuelve como valor dicho resultado.

En este ejemplo hemos utilizado otra cualidad de las funciones como ciudadanos de primera clase y es que se le puede inspecionar por la variable especial `__name__` que nos devuelve un una cadena con el nombre original de la función.

Este ejemplo ha sido solo con el objetivo de ilustrar de manera simple que una función se puede pasar como argumento a otra. Puede parecerle de poca utilidad al lector ya que si lo que queremos es el resultado de aplicar una función *`f`* a un valor *`x`* no necesitamos de una tal funcion *`convierte`* porque bastaría con escribir directamente *`f(x)`* allí donde nos interesa el resultado de aplicar *`f`* . Más adelante veremos como esta capacidad nos permite lograr construcciones de mayor utilidad como definir *funciones que devuelven funciones*  

Observe también como un objeto función puede *inspeccionarse*. La implementación de la función `covierte(f,valor)` del código a continuación recibe un parámetro función en `f` y le pide el nombre con la variable especial `__name__` que tiene el nombre de la función

```python
#FUNCIONES DE CONVERSION
def kg_a_libras(kg):
    """Convierte kilogramos a libras."""
    return kg * 2.20462262

def libra_a_kgs(libra):
    """Convierte libras a kilogramos."""
    return libra / 2.20462262

def cm_a_pulgadas(cm):
    """Convierte centímetros a pulgadas."""
    return cm / 2.54

def pulgada_a_cms(pulgadas):
    """Convierte pulgadas a centímetros."""
    return pulgadas * 2.54

def celsius_a_fahrenheit(celsius):
    """Convierte grados Celsius a Fahrenheit."""
    return (celsius * 9/5) + 32

def fahrenheit_a_celsius(fahrenheit):
    """Convierte grados Fahrenheit a Celsius."""
    return (fahrenheit - 32) * 5/9

def convierte(f, valor):
    result = f(valor)
    print(f'\nAplicar {f.__name__} a {valor_a_convertir} nos da {result}') #la variable especial __name__ forma parte de todo objeto que sea función 
    return result

valor_a_convertir = 100
result = convierte(kg_a_libras, valor_a_convertir)
print(f' El resultado + 1 es {result+1}')

result = convierte(pulgada_a_cms, valor_a_convertir)
print(f' El resultado + 1 es {result+1}')

```

## 7.4 Valores función dentro de estructuras de datos
Al ser las funciones objetos que pueden tratarse como tales pueden almacenarse dentro de estructuras como listas, tuplas, conjuntos o diccionarios o como valores de variables de otros objetos.

Considere las siguientes funciones de conversión anteriores `kg_a_libras`, `libra_a_kgs`, `cm_a_pulgadas`, `pulgada_a_cms`, `celsius_a_fahrenheit`, `fahrenheit_a_celsius` convierten una unidad de medida de un sistema a su correspondiente en otro sistema.

**Opcion1 Tradicional en Programación imperativa** 
El código de la *Opcion1* a continuación puede formar parte de una aplicación que quiere convertir una unidad a otra según lo que indique el usuario. La interacción para la entrada de los datos se ha simplificado para facilitar la comprensión del ejemplo.

Esta opción tiene el inconveniente la larga secuencia de **if** y **elif** que invoca directamente a la función corresondiente a la pregunta **if** Esto dificulta la comprensión del código. Además si se quiere añadir otra conversión habría que hacer cambios en esta secuencia de **if** 

**Opcion2 Programación Funcional** 
En esta opción los objetos función se guardarán como valores en un *diccionario*. 

Esta opción usa un diccionario que tiene como llaves (claves) el nombre de la función que da el usuario en la interacción se utiliza como nombre de lo que quiere hacer y tiene como valor asociado a dicha llave la función que quiere aplicar.

Además de ser más simple y concisa de leer, esta alternativa es más **MANTENIBLE**. Si más adelante se quiere añadir una nueva conversión (por ejemplo, *kilómetros a millas*), solo hay que escribir la funcion correspondiente y poner el par *llave-valor* en el diccionario pero no hay que hacer modificaciones en el "chorizo" de *if-elif* de los bloques largos de código condicional.

```python
#OPCION 1 IMPERATIVA TRADICIONAL
# Entrada de la conversión a realizar.
# esto realmente pudiera ser mas realista según la aplicación real que interactua con el usuario
conversion = input("¿Qué conversión deseas hacer? (ej: kg a libras, cm a pulgadas, celsius a fahrenheit...): ").lower().strip()

#Entrada para el valor numérico (asegurando convertirlo a float)
valor_a_convertir = float(input("Introduce el valor a convertir: "))

#Decidir cuál conversión aplicar usando una secuencia de if / elif
if conversion == "kg a libras" or conversion == "kilos a libras":
    resultado = kg_a_libras(valor_a_convertir)
    print(f"Resultado: {valor_a_convertir} kg equivalen a {resultado:.2f} libras")

elif conversion == "libras a kg" or conversion == "libras a kilos":
    resultado = libra_a_kgs(valor_a_convertir)
    print(f"Resultado: {valor_a_convertir} libras equivalen a {resultado:.2f} kg")

elif conversion == "cm a pulgadas":
    resultado = cm_a_pulgadas(valor_a_convertir)
    print(f"Resultado: {valor_a_convertir} cm equivalen a {resultado:.2f} pulgadas")

elif conversion == "pulgadas a cm":
    resultado = pulgada_a_cms(valor_a_convertir)
    print(f"Resultado: {valor_a_convertir} pulgadas equivalen a {resultado:.2f} cm")

elif conversion == "celsius a fahrenheit":
    resultado = celsius_a_fahrenheit(valor_a_convertir)
    print(f"Resultado: {valor_a_convertir} °C equivalen a {resultado:.2f} °F")

elif conversion == "fahrenheit a celsius":
    resultado = fahrenheit_a_celsius(valor_a_convertir)
    print(f"Resultado: {valor_a_convertir} °F equivalen a {resultado:.2f} °C")

else:
    print("Esta conversión no existe, verifica bien lo que me has dado.")

#Ejecuta este código probando diferentes opciones
```

```python
# OPCION 2. PROGRAMACION FUNCIONAL
conversiones = {
    "kg a libras": kg_a_libras,
    "libra a kgs": libra_a_kgs,
    "cm a pulgadas": cm_a_pulgadas,
    "pulgada a cms": pulgada_a_cms,
    "celsius a fahrenheit": celsius_a_fahrenheit,
    "fahrenheit a celsius": fahrenheit_a_celsius
}

#Dialogo con usuario
conversion = input("¿Qué conversión deseas hacer? (ej: kg a libras, cm a pulgadas...): ").lower()
valor_a_convertir = float(input("Introduce el valor a convertir: "))

if conversion in conversiones:
    # Extraemos la función del diccionario usando la clave y se ejecutamos pasándole el valor
    print(f"Resultado: {valor_a_convertir} covertido es -> {conversiones[conversion](valor_a_convertir) :.2f}")
else:
    print("Esta conversión no existe, verifica bien lo que me has dado.")
```

## 7.5 Funcion que devuelve una función
Otra ventaja de poder tratar a una función como ciudadano de primera clase es que *una función puede devolver otra función como resultado de su ejecución*

La función `crear_multiplicador` del código siguiente define internamente una función `multiplicar` que multiplica el argumento (`valor`) que se le pase por el (`factor`) conque fue creada cuando se llamó a `crear_multiplicador(factor)`

Quiere decir que cuando se hace

`duplicar = crear_multiplicador(2)` lo que se le está asignando a `duplicar` es la función

    def multiplicar(valor):
        return valor * 2

creada internamente y devuelta cuando se ejecutó la llamada. De modo que si ahora se invoca a `duplicar(101)` lo que este retorna es el valor de ejecutar

`101*2`

```python
def crear_multiplicador(factor):

    def multiplicar(valor):
        return valor * factor

    return multiplicar

duplicar = crear_multiplicador(2)
triplicar = crear_multiplicador(3)
quintuplicar = crear_multiplicador(5)

valor = 101
print(f'{valor} --> {duplicar(valor)}')
print(f'{valor} --> {triplicar(valor)}')
print(f'{valor} --> {quintuplicar(valor)}')
```

## 7.6 Función de transformación que recibe una función y devuelve una función
Usted puede pensar, y con razón, que el ejemplo anterior de `multiplicador` es realmente *mucho ruido y pocas nueces*. Para este ejemplo bastaría que allí donde se quería duplicar un `valor` haber escrito directamente `valor*2` sin tanta parafernalia.

Sin embargo, veamos un caso con más sustancia. La siguiente función `medible` recibe como argumento una función `f_a_medir` pero devuelve otra función que hace lo mismo que `f_a_medir` pero que también devuelve el valor que hubiese devuelto la función `f_a_medir` original y además el devuelve el tiempo que demoró en ejecutar la misma.

La función que hemos llamado `medible` define internamente otra función `f_medible(*args)`. Recuerde que al usar con argumento `*args`estamos diciendo que la función `f_medible` se puede llamar con cualquier cantidad de parámetros. Note que esta función `f_medible` es la que mandará a ejecutar la llamada `f_a_medir(*args)` y medir el tiempo de ejecutar la misma. Es esta función interna la que se devolverá como resultado de le llamada a `medible(f_a_medir)`.

Cuando en el ejemplo a continuación se hace

`a_mayuscula = medible(to_upper_lento)`

se le está asignando como valor a la variable `a_mayuscula` la función retornada por la llamada `medible(to_upper_lento)`

Cuando se hace una llamada a través de dicha variable, como cuando en este ejemplo se hace `a_mayuscula(nombre)`, es que realmente se ejecutará, midiendo el tiempo que demora, la llamada a `to_upper_lento(nombre)`. Como se ha llamado con un solo argumento (en este caso la variable `nombre` que tiene como valor la cadena `"juana"`) es este valor el que le pasa como único valor al parámetro `*args`. De este modo usando el recurso `args` la función `medible` crea funciones que se pueden aplicar a los mismos parámeros que le aplicarían a la función original


```python
import time
def medible(f_a_medir):
  #Esta función devuelve una función que lo que hace es devolver un tuplo de dos valores, uno es el resultado de evaluar func_a_medir(args) y el otro es el tiempo que demoró en ejecutarlo
  def f_medible(*args):
    inicio = time.perf_counter()
    result = f_a_medir(*args)
    tiempo = time.perf_counter() - inicio
    return result, tiempo
  return f_medible

def to_upper_lento(texto:str) -> str: #Simular una funcion que demora en convertir a mayúscula
  #Código que procesa los argumentos
  time.sleep(len(texto)) #Demorar la conversión según la longitud del texto
  return texto.upper()

a_mayuscula = medible(to_upper_lento)
nombre = "juana"
resultado, t = a_mayuscula(nombre)
print(f'Nombre original {nombre} convertido a mayúsculas {resultado} ha demorado {t:.4f} segs')
```

```python
## RESPUESTA AL EJERCICIO 2. NO MIRE ESTE CÓDIGO ANTES DE INTENTAR DE RESOLVER POR SU CUENTA
import time
def medible(f_a_medir):
  #Esta función devuelve una función que lo que hace es devolver un tuplo de dos valores, uno es el resultado de evaluar func_a_medir(args) y el otro es el tiempo que demoró en ejecutarlo
  def f_medible(*args):
    inicio = time.perf_counter()
    result = f_a_medir(*args)
    tiempo = time.perf_counter() - inicio
    return result, tiempo
  return f_medible

def fib_iterativo(n: int) -> int:
    if n < 0:
        raise ValueError("El índice 'n' debe ser un número entero no negativo.")
    if n == 0:
        return 0
    if n == 1:
        return 1
    penultimo, ultimo = 0, 1   
    for _ in range(2, n + 1):
        penultimo, ultimo = ultimo, penultimo + ultimo        
    return ultimo

def fib_recursivo(n: int) -> int:
    # RECUERDE QUE ESTA SOLUCION ES MUY INEFICIENTE PORQUE HACE CALCULOS REPETIDOS (O(2^n)).  
    if n < 0:
        raise ValueError("El índice 'n' debe ser un número entero no negativo.")
    # Casos base
    if n == 0:
        return 0
    if n == 1:
        return 1    
    # Llamada recursiva que repite muchos cálculos
    return fib_recursivo(n - 1) + fib_recursivo(n - 2)

fib_eficiente = medible(fib_iterativo)
fib_ineficiente = medible(fib_recursivo)
numero = 40
resultado, t = fib_eficiente(numero)
print(f'Fibonnaci eficiente iterativo de   {numero} es {resultado} calculado en {t:.4f} segs')
resultado, t = fib_ineficiente(numero)
print(f'Fibonnaci recursivo ineficiente de {numero} es {resultado} calculado en {t:.4f} segs')
```
